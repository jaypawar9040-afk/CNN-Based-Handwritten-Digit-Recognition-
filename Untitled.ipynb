```python
X_train = X_train.astype("float32")/255.0
X_test = X_test.astype("float32")/255.0
# X_train[0]
```

### 3. Reshape data to include channel dimantion (since CNN expects 4D input)

```python
X_train = X_train.reshape(-1,28,28,1)
X_test = X_test.reshape(-1,28,28,1)
```

## Now we will move ahead to Build CNN
<img src="CNN image.jpg.jpeg" width="1000">


```python
from tensorflow.keras import models, layers

model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),

    # First Convolutional Layer
    layers.Conv2D(32, (3, 3), activation="relu"),

    # First Max Pooling Layer
    layers.MaxPooling2D((2, 2)),

    # Second Convolutional Layer
    layers.Conv2D(64, (3, 3), activation="relu"),

    # Second Max Pooling Layer
    layers.MaxPooling2D((2, 2)),

    # Flatten Layer
    layers.Flatten(),

    # Fully Connected Layer
    layers.Dense(64, activation="relu"),

    # Output Layer (0-9)
    layers.Dense(10, activation="softmax")
])

model.summary()
# In above model we have build with 7 layers
```

```python
model.compile(optimizer = "adam", loss = "sparse_categorical_crossentropy", metrics = ['accuracy'])
```

```python
history = model.fit(X_train, y_train, epochs = 5, batch_size = 512 , verbose = 1, validation_split = 0.2)
```

```python
plt.plot(history.history['loss'], color = 'red', label = 'Training loss') 
plt.plot(history.history['val_loss'], color = 'green', label = 'Validation loss') 
plt.legend() 
plt.show()
```

```python
plt.plot(history.history['accuracy'], color = 'orange', label = 'Training accuracy') 
plt.plot(history.history['val_accuracy'], color = 'blue', label = 'Validation Accuracy') 
plt.legend() 
plt.show()
```

```python
from PIL import Image, ImageOps
import numpy as np

def pred_for(img_path):

    img = Image.open(img_path).convert("L")

    # Convert black digit on white background
    img = ImageOps.invert(img)

    # Crop unnecessary background
    bbox = img.getbbox()
    if bbox is None:
        raise ValueError("No digit found in image")

    img = img.crop(bbox)

    # Maintain aspect ratio
    img.thumbnail((20, 20))

    # Place digit in a 28x28 black image
    canvas = Image.new("L", (28, 28), 0)

    x = (28 - img.width) // 2
    y = (28 - img.height) // 2

    canvas.paste(img, (x, y))

    # Normalize and reshape
    img_array = np.array(canvas).astype("float32") / 255.0
    img_array = img_array.reshape(1, 28, 28, 1)

    # Predict
    prediction = model.predict(img_array, verbose=0)

    return np.argmax(prediction)
```

```python
print("Predicted Digit:", pred_for("Bold Black Number 7.png"))
```
